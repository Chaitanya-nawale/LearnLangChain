## **Default Chat Model Setup**

```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

google_model="gemini-3.1-flash-lite"
```

## **Fundamental Process**

```python
# Task 1 : Prompt Generation with Dynamic Inputs
from langchain_core.prompts import ChatPromptTemplate

dynamic_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a genz assistant. You are very chill and use a lot of slang. You are also very funny and sarcastic."),
    ("user", "Write a joke about {topic}.")
])
```

```python
# Task 2 : LLM
from langchain.chat_models import init_chat_model

llm_gemini = init_chat_model(model=google_model,model_provider="openai",
    openai_api_key=os.environ["GOOGLE_API_KEY"],
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
# Task 3 : String Output Parser
from langchain_core.output_parsers import StrOutputParser

str_output_parser = StrOutputParser()
```

## **Manual Invocation**

```python

ready_prompt = dynamic_prompt.invoke({"topic": input("Enter a topic for the joke: ")})

response = llm_gemini.invoke(ready_prompt)

str_output_parser.invoke(response)
```

## **LangChain using Pipe Operator**

```python
chain = dynamic_prompt | llm_gemini | str_output_parser

chain.invoke({"topic": input("Enter a topic for the joke: ")})
```

## **LangChain using Runnable**

```python
from langchain_core.runnables import RunnableSequence

runnable_chain = RunnableSequence(dynamic_prompt, llm_gemini, str_output_parser)

runnable_chain.invoke({"topic": input("Enter a topic for the joke: ")})
```
