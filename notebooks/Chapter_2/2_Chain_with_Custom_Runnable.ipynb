## **Default Chat Model Setup**

```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

google_model="gemini-3.1-flash-lite"
```

## **Fundamental Process**

```python
# Task 1 : Prompt Generation with Dynamic Inputs
from langchain_core.prompts import ChatPromptTemplate

dynamic_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a genz assistant. You are very chill and use a lot of slang. You are also very funny and sarcastic."),
    ("user", "Write a joke about {topic}.")
])
```

```python
# Task 2 : LLM
from langchain.chat_models import init_chat_model

llm_gemini = init_chat_model(model=google_model,model_provider="openai",
    openai_api_key=os.environ["GOOGLE_API_KEY"],
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
# Task 3 : String Output Parser
from langchain_core.output_parsers import StrOutputParser

str_output_parser = StrOutputParser()
```

```python
# Task 4 : Runnable Function
from langchain_core.runnables import RunnableLambda

def joke_dict(joke: str) -> dict:
    return {"joke": joke}

joke_runnable = RunnableLambda(joke_dict)
```

```python
# Task 5 : Prompt Generation with Dynamic Inputs

dynamic_prompt_analyser = ChatPromptTemplate.from_messages([
    ("system", "You are a joke analyser. Analyse the joke and return if it is funny or not. Provide the joke as well as a feedback on how to improve the joke if it is not funny."),
    ("user", "{joke}")
])
```

```python
# Task 6 : LLM
llm_gemini
```

```python
#Task 7 :  String Output Parser
str_output_parser
```

## **Custom Runnable Chain**

```python
chain = dynamic_prompt | llm_gemini | str_output_parser | joke_runnable | dynamic_prompt_analyser | llm_gemini | str_output_parser
```

```python
chain.invoke({"topic": input("Enter a topic for the joke: ")})
```
