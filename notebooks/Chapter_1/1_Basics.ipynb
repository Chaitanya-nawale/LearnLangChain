## **Default Chat Model Setup**

```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

google_model="gemini-3.1-flash-lite"
```

```python
from openai import OpenAI

# Initialize the OpenAI client with Gemini's endpoint
client = OpenAI(
    api_key=os.environ.get("GOOGLE_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# Call the Gemini model using OpenAI style Chat Completions
response = client.chat.completions.create(
    model=google_model,  # Specify your target Gemini model
    messages=[
        {"role": "system", "content": "You are a helpful coding assistant."},
        {"role": "user", "content": "Tell me a joke."}
    ],
    temperature=0.7,
    max_tokens=150
)

# Extract and print the content exactly like an OpenAI response
print(response.choices[0].message.content)
```

## **Using LangChain Chat Model**

```python
# Approach using LangChain's Google wrapper for Gemini's endpoint

# from langchain_google_genai import ChatGoogleGenerativeAI

# llm_gemini = ChatGoogleGenerativeAI(model=google_model, temperature=0)

# llm_gemini.invoke("Tell me a joke").content

# Alternative approach using LangChain's OpenAI wrapper for Gemini's endpoint
import os
from langchain_openai import ChatOpenAI

# Use the ChatOpenAI wrapper pointing to Gemini's OpenAI-compatible endpoint
llm_gemini = ChatOpenAI(
    model=google_model,
    openai_api_key=os.environ["GOOGLE_API_KEY"],
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# This will return a clean OpenAI-formatted AIMessage
llm_gemini.invoke("Tell me a joke")
```

```python
from langchain.chat_models import init_chat_model

# llm_gemini = init_chat_model(model=google_model,model_provider="google_genai")
llm_gemini = init_chat_model(model=google_model,model_provider="openai",
    openai_api_key=os.environ["GOOGLE_API_KEY"],
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
llm_gemini.invoke("Tell me a joke")
```
