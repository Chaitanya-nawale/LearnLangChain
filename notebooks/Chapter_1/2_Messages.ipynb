## **Default Chat Model Setup**

```python
from dotenv import load_dotenv
load_dotenv()

import getpass
import os

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Groq API key: ")

google_model="gemini-3.1-flash-lite"
```

```python
from langchain.chat_models import init_chat_model

llm_gemini = init_chat_model(model=google_model,model_provider="openai",
    openai_api_key=os.environ["GOOGLE_API_KEY"],
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

## **Messages**

```python
from langchain_core.messages import HumanMessage, SystemMessage

my_messages = [
    SystemMessage("You are a genz assistant. You are very chill and use a lot of slang. You are also very funny and sarcastic."),
    HumanMessage("Hello! Tell me a joke about programming."),
]

llm_gemini.invoke(my_messages).content
```

## **Prompts**

```python
from langchain_core.prompts import PromptTemplate

input_topic = input("Enter a topic: ")

dynamic_prompt = PromptTemplate.from_template("Write a joke about {topic} in a genz tone.")

ready_prompt = dynamic_prompt.format(topic=input_topic)

llm_gemini.invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

input_topic = input("Enter a topic: ")
input_tone = input("Enter a tone: ")

dynamic_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a {tone} assistant. You are very chill and use a lot of slang. You are also very funny and sarcastic."),
    ("user", "Write a joke about {topic} in a {tone} tone.")
])

ready_prompt = dynamic_prompt.invoke({"topic": input_topic, "tone": input_tone})

ready_prompt
```
